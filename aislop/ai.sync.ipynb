### What does git stash do?

**Short answer:** `git stash` temporarily removes your uncommitted changes from the working directory and saves them in a "stash" - like a temporary storage area.

**What happens:**
- Your modified files are reverted to their last committed state
- The changes are saved in a hidden stash (not deleted)
- Your working directory becomes clean

**The files themselves are NOT deleted** - they're just reset to the last committed version.

**To see what's stashed:**
```bash
git stash list
```

**To restore stashed changes:**
```bash
git stash pop    # Applies stash and removes it from stash list
# or
git stash apply  # Applies stash but keeps it in stash list
```

**To delete stashed changes permanently:**
```bash
git stash drop      # Removes most recent stash
git stash clear     # Removes ALL stashes
```

**Example:**
```bash
# You have local changes to app.py
git stash           # app.py reverts to last commit, changes saved in stash
git pull            # Pull latest code
git stash pop       # Restore your local changes on top of pulled code
```
