### Installing Miniconda on a Remote Server
1. **Download Miniconda installer:**
   
For Linux (64-bit):
```bash
wget https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh
```
2. **Run the installer:**

```bash
bash Miniconda3-latest-Linux-x86_64.sh
```
- Follow the prompts (accept license, choose install location, etc.).
3. **Initialize conda:**

```bash
~/miniconda3/bin/conda init
```
- Restart your shell or run:
```bash
source ~/.bashrc
```
4. **Verify installation:**

```bash
conda --version
```
5. **(Optional) Remove installer:**
   
   ```bash
   rm Miniconda3-latest-Linux-x86_64.sh
   ```
You can now use `conda` to create or update environments on the remote server.

How do I make all the folders and files in /home to be owned byy abv

To change ownership of all files and folders in /home to the user `abv`, use the following command as root (or with sudo):
```bash
sudo chown -R abv:abv /home
```
- `-R` means recursive (applies to all subfolders and files).
- `abv:abv` sets both user and group ownership to `abv`.
**Warning:** This will change ownership for everything in /home, which may affect other users or system files. Use with caution.

To replicate your local conda environment on a remote server, follow these steps:

1. On your local machine, export the environment:
conda env export --no-builds > environment.yml

2. Transfer the environment.yml file to the remote server (e.g., using scp or another file transfer method).

3. On the remote server, create the environment from the file:
conda env create -f environment.yml

If you want to update an existing environment on the remote, use:
conda env update -f environment.yml

### duplicationg conda environment onto server
```bash
conda env export --no-builds > environment.yml

# Copy environment.yml from local to remote server:
scp environment.yml username@remote_host:/path/to/destination/

# Example:
scp environment.yml abv@your.server.com:/home/abv/

# To copy from remote to local, reverse the order:
scp username@remote_host:/path/to/environment.yml ./
```

Yes, whenever you change the local conda environment (add, remove, or update packages), you should re-export the environment.yml, transfer it to the server, and update the environment there.

Steps:
1. Re-export:
   ```bash
   conda env export --no-builds > environment.yml
   ```
2. Transfer:
   ```bash
   scp environment.yml username@remote_host:/path/to/destination/
   ```
3. On the server, update the environment:
   ```bash
   conda env update -f environment.yml
   ```
This keeps the remote environment in sync with your local changes.

To export your conda environment without including pip packages, use:
```bash
conda env export --no-builds | grep -v '^- pip:' > environment.yml
```
This command filters out the pip section from the exported environment.yml file.

To keep only a specific pip package (e.g., `abv==0.1.0`) in your environment.yml, you can manually edit the file after export:
1. Export as usual:
   ```bash
   conda env export --no-builds > environment.yml
   ```
2. Open environment.yml in a text editor.
3. In the `- pip:` section, delete all lines except:
   ```yaml
   - pip:
     - abv==0.1.0
   ```
This will ensure only `abv==0.1.0` is included from pip when recreating the environment.
