## Solution: Jupyter Notebook Variable Redeclaration Issue

**Problem**: `var` declarations in Jupyter notebooks persist across cell executions, causing "already been declared" errors when re-running cells.

**Root Cause**: JavaScript cells in Jupyter share a **global scope**. When you re-run a cell with `var normalizeFilePath`, it tries to redeclare an already existing variable.

**Solutions**:

`````javascript
// ❌ PROBLEMATIC: Using var in Jupyter (causes redeclaration errors)
// var normalizeFilePath = (originalName) => { ... }  // Error on re-run!

// ✅ SOLUTION 1: Use let/const instead of var (recommended)
const normalizeFilePath = (originalName) => {
    const rawPath = String.raw`${originalName}`;
    return rawPath.replace(/\\/g, '/');
}

const path = String.raw`C:\Users\mcken\OneDrive\books\NGINX_Cookbook_Part1.pdf`;
const npath = normalizeFilePath(path);
console.log('Normalized path:', npath);

// Note: Even const/let can cause errors on re-run, but they're clearer about the issue
`````

`````javascript
// ✅ SOLUTION 2: Use function declarations (they can be redeclared)
function normalizeFilePath2(originalName) {
    const rawPath = String.raw`${originalName}`;
    return rawPath.replace(/\\/g, '/');
}

// Functions can be safely redeclared in Jupyter
const testPath = String.raw`C:\Users\mcken\OneDrive\books\NGINX_Cookbook_Part1.pdf`;
console.log('Function approach:', normalizeFilePath2(testPath));
`````

`````javascript
// ✅ SOLUTION 3: Use immediately invoked function expression (IIFE)
// This creates a local scope and avoids global variable conflicts
(() => {
    const normalizeFilePath = (originalName) => {
        const rawPath = String.raw`${originalName}`;
        return rawPath.replace(/\\/g, '/');
    };
    
    const path = String.raw`C:\Users\mcken\OneDrive\books\NGINX_Cookbook_Part1.pdf`;
    const npath = normalizeFilePath(path);
    console.log('IIFE approach:', npath);
})();

// Variables inside IIFE don't pollute global scope
`````

`````javascript
// 2. ENHANCED uploadFileToServer() - Handle conversion responses
async uploadFileToServer(file, fileData) {
    const formData = new FormData();
    formData.append('file', file);
    formData.append('cleanedName', fileData.cleanedName);
    formData.append('fullPath', fileData.fullPath);
    formData.append('originalName', fileData.originalName);
    formData.append('folder', fileData.fullPath.split('/').slice(0, -1).join('/'));
    
    // Check if this file will be converted
    const isDocx = file.name.toLowerCase().endsWith('.docx');
    if (isDocx) {
        console.log('📄 .docx file detected - server will convert to PDF');
    }
    
    const response = await fetch(`${this.songManager.API_BASE}/api/upload`, {
        method: 'POST',
        body: formData
    });
    
    if (!response.ok) {
        throw new Error(`Upload failed: ${response.statusText}`);
    }
    
    const result = await response.json();
    
    // Handle conversion response
    if (result.converted) {
        console.log(`🔄 File converted: ${result.original_filename} → ${result.filename}`);
        console.log(`📂 Stored as: ${result.path}`);
        
        // Update fileData with converted info
        fileData.cleanedName = result.filename;
        fileData.fullPath = result.path;
        fileData.wasConverted = true;
        fileData.conversionType = result.conversion_type;
        fileData.conversionMessage = result.message;
    }
    
    console.log('✅ Upload successful:', result);
    return result;
}
`````
