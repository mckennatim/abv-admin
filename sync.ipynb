# sync
* no original work gets done on server. 
* code in admin can reach into /home/abv to change static files and create new html there. 
* once code is tested in WSL, actual change to site happens from server. 
* test changes happen in WSL, but they are ignored
* WSL sessions still start with rsnc abv->vivo
* abv is the source of truth for the static html pages and /resources. Before you do anything on dev, get current copies of all this isung rsync




## sync abv -> dev
Do this before you start working on dev
### rsync
```bash
rsync -avz abv@abvchorus.org:/home/abv/public_html/*.html /home/abv/public_html/

# resources: abv->dev Delete on vivo if not on abv
rsync -avz --delete abv@abvchorus.org:/home/abv/public_html/resources /home/abv/public_html/

# assets: abv->dev leave test data on dev, reproduce it manually on abv
rsync -avz abv@abvchorus.org:/home/abv/public_html/assets /home/abv/public_html/
```
### git
```bash
sudo chown -R abv:abv abv-admin
# from public_html/abv-admin
git add . -A
git commit -m"xxxx"
git push -u origin main
```
on the server `git pull`

### update conda on server
```bash
# on local
cd /home/abv/public_html
conda env export --no-builds > environment.yml
# EDIT environment.yml remove the line      - abv==0.1.0
scp environment.yml abv@abvchorus.org:/home/
# on remote
cd /home
conda env update -f environment.yml

find /home/abv -name "songs.db"
```
