### untracking git
and its consequences.

If you are getting stash errors on ABV git pull, you may want to stop tracking those files on WSL. That should get rid of the stash errors on git pull.

| Command | Effect |
|---------|--------|
| `git rm --cached nohup.out` | Stop tracking, **keep** file locally |
| `git rm nohup.out` | Stop tracking, **delete** file locally |
| `echo 'nohup.out' >> .gitignore` | Prevent future tracking |
| `git add .gitignore && git commit` | Save changes |

### update conda on server
You will want to update conda on abv whenever you add libraries to wsl.
```bash
# on local
cd /home/abv/public_html
conda env export --no-builds > environment.yml
# EDIT environment.yml remove the line      - abv==0.1.0
scp environment.yml abv@abvchorus.org:/home/
# on remote
cd /home
conda env update -f environment.yml
```
